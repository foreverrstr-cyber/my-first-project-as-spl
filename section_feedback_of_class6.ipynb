```python
# ----------------------------------------------------
# PROJECT: School Section Performance Tracker (Named Categories Edition)
# ----------------------------------------------------

sections_data = {
    "m1": {
        "name": "M1",
        "discipline": 85,
        "speaking": 70,
        "writing": 90,
        "academic": 80,
        "phrase": "You have a nice average score, keep it up!",
        "special": "The best discipline class is M1!"
    },
    "f1": {
        "name": "F1",
        "discipline": 80,
        "speaking": 70,
        "writing": 80,
        "academic": 75,
        "phrase": "You have the fire, take it out and keep it up!",
        "special": "Your section has nice potential, keep it up!"
    },
    "o1": {
        "name": "O1",
        "discipline": 75,
        "speaking": 90,
        "writing": 80,
        "academic": 95,
        "phrase": "Totally your section is super competitive!",
        "special": "You are the best speakers' section!"
    },
    "o2": {
        "name": "O2",
        "discipline": 80,
        "speaking": 80,
        "writing": 90,
        "academic": 90,
        "phrase": "Your section is perfect!",
        "special": "You are the best handwriting section, keep the elegance!"
    }
}

print("==========================================")
print("🏆 SCHOOL SECTION PERFORMANCE TRACKER 🏆")
print("==========================================")

user_section = input("Which section are you from? (M1, F1, O1, O2): ").strip().lower()

if user_section in sections_data:
    data = sections_data[user_section]

    # MATH LOGIC: Collect the individual scores into a list temporarily to calculate the average
    all_scores = [data["discipline"], data["speaking"], data["writing"], data["academic"]]
    calculated_avg = sum(all_scores) / len(all_scores)

    print(f"\n📊 Performance Report for Section {data['name']}:")
    print(f"• Discipline         : {data['discipline']}/100")
    print(f"• Speaking Skills    : {data['speaking']}/100")
    print(f"• Writing Skills     : {data['writing']}/100")
    print(f"• Academic Score     : {data['academic']}/100")
    print(f"• Total Average      : {calculated_avg:.1f}/100")  # Automatically calculated math!

    print("\n💬 Class Feedback:")
    print(f"- {data['phrase']}")
    print(f"- {data['special']}")
else:
    print("❌ Invalid section name! Please choose from M1, F1, O1, or O2.")# ----------------------------------------------------
# PROJECT: School Section Performance Tracker (Named Categories Edition)
# ----------------------------------------------------

sections_data = {
    "m1": {
        "name": "M1",
        "discipline": 85,
        "speaking": 70,
        "writing": 90,
        "academic": 80,
        "phrase": "You have a nice average score, keep it up!",
        "special": "The best discipline class is M1!"
    },
    "f1": {
        "name": "F1",
        "discipline": 80,
        "speaking": 70,
        "writing": 80,
        "academic": 75,
        "phrase": "You have the fire, take it out and keep it up!",
        "special": "Your section has nice potential, keep it up!"
    },
    "o1": {
        "name": "O1",
        "discipline": 75,
        "speaking": 90,
        "writing": 80,
        "academic": 95,
        "phrase": "Totally your section is super competitive!",
        "special": "You are the best speakers' section!"
    },
    "o2": {
        "name": "O2",
        "discipline": 80,
        "speaking": 80,
        "writing": 90,
        "academic": 90,
        "phrase": "Your section is perfect!",
        "special": "You are the best handwriting section, keep the elegance!"
    }
}

print("==========================================")
print("🏆 SCHOOL SECTION PERFORMANCE TRACKER 🏆")
print("==========================================")

user_section = input("Which section are you from? (M1, F1, O1, O2): ").strip().lower()

if user_section in sections_data:
    data = sections_data[user_section]

    # MATH LOGIC: Collect the individual scores into a list temporarily to calculate the average
    all_scores = [data["discipline"], data["speaking"], data["writing"], data["academic"]]
    calculated_avg = sum(all_scores) / len(all_scores)

    print(f"\n📊 Performance Report for Section {data['name']}:")
    print(f"• Discipline         : {data['discipline']}/100")
    print(f"• Speaking Skills    : {data['speaking']}/100")
    print(f"• Writing Skills     : {data['writing']}/100")
    print(f"• Academic Score     : {data['academic']}/100")
    print(f"• Total Average      : {calculated_avg:.1f}/100")  # Automatically calculated math!

    print("\n💬 Class Feedback:")
    print(f"- {data['phrase']}")
    print(f"- {data['special']}")
else:
    print("❌ Invalid section name! Please choose from M1, F1, O1, or O2.")# ----------------------------------------------------
# PROJECT: School Section Performance Tracker (Named Categories Edition)
# ----------------------------------------------------

sections_data = {
    "m1": {
        "name": "M1",
        "discipline": 85,
        "speaking": 70,
        "writing": 90,
        "academic": 80,
        "phrase": "You have a nice average score, keep it up!",
        "special": "The best discipline class is M1!"
    },
    "f1": {
        "name": "F1",
        "discipline": 80,
        "speaking": 70,
        "writing": 80,
        "academic": 75,
        "phrase": "You have the fire, take it out and keep it up!",
        "special": "Your section has nice potential, keep it up!"
    },
    "o1": {
        "name": "O1",
        "discipline": 75,
        "speaking": 90,
        "writing": 80,
        "academic": 95,
        "phrase": "Totally your section is super competitive!",
        "special": "You are the best speakers' section!"
    },
    "o2": {
        "name": "O2",
        "discipline": 80,
        "speaking": 80,
        "writing": 90,
        "academic": 90,
        "phrase": "Your section is perfect!",
        "special": "You are the best handwriting section, keep the elegance!"
    }
}

print("==========================================")
print("🏆 SCHOOL SECTION PERFORMANCE TRACKER 🏆")
print("==========================================")

user_section = input("Which section are you from? (M1, F1, O1, O2): ").strip().lower()

if user_section in sections_data:
    data = sections_data[user_section]

    # MATH LOGIC: Collect the individual scores into a list temporarily to calculate the average
    all_scores = [data["discipline"], data["speaking"], data["writing"], data["academic"]]
    calculated_avg = sum(all_scores) / len(all_scores)

    print(f"\n📊 Performance Report for Section {data['name']}:")
    print(f"• Discipline         : {data['discipline']}/100")
    print(f"• Speaking Skills    : {data['speaking']}/100")
    print(f"• Writing Skills     : {data['writing']}/100")
    print(f"• Academic Score     : {data['academic']}/100")
    print(f"• Total Average      : {calculated_avg:.1f}/100")  # Automatically calculated math!

    print("\n💬 Class Feedback:")
    print(f"- {data['phrase']}")
    print(f"- {data['special']}")
else:
    print("❌ Invalid section name! Please choose from M1, F1, O1, or O2.")# ----------------------------------------------------
# PROJECT: School Section Performance Tracker (Named Categories Edition)
# ----------------------------------------------------

sections_data = {
    "m1": {
        "name": "M1",
        "discipline": 85,
        "speaking": 70,
        "writing": 90,
        "academic": 80,
        "phrase": "You have a nice average score, keep it up!",
        "special": "The best discipline class is M1!"
    },
    "f1": {
        "name": "F1",
        "discipline": 80,
        "speaking": 70,
        "writing": 80,
        "academic": 75,
        "phrase": "You have the fire, take it out and keep it up!",
        "special": "Your section has nice potential, keep it up!"
    },
    "o1": {
        "name": "O1",
        "discipline": 75,
        "speaking": 90,
        "writing": 80,
        "academic": 95,
        "phrase": "Totally your section is super competitive!",
        "special": "You are the best speakers' section!"
    },
    "o2": {
        "name": "O2",
        "discipline": 80,
        "speaking": 80,
        "writing": 90,
        "academic": 90,
        "phrase": "Your section is perfect!",
        "special": "You are the best handwriting section, keep the elegance!"
    }
}

print("==========================================")
print("🏆 SCHOOL SECTION PERFORMANCE TRACKER 🏆")
print("==========================================")

user_section = input("Which section are you from? (M1, F1, O1, O2): ").strip().lower()

if user_section in sections_data:
    data = sections_data[user_section]

    # MATH LOGIC: Collect the individual scores into a list temporarily to calculate the average
    all_scores = [data["discipline"], data["speaking"], data["writing"], data["academic"]]
    calculated_avg = sum(all_scores) / len(all_scores)

    print(f"\n📊 Performance Report for Section {data['name']}:")
    print(f"• Discipline         : {data['discipline']}/100")
    print(f"• Speaking Skills    : {data['speaking']}/100")
    print(f"• Writing Skills     : {data['writing']}/100")
    print(f"• Academic Score     : {data['academic']}/100")
    print(f"• Total Average      : {calculated_avg:.1f}/100")  # Automatically calculated math!

    print("\n💬 Class Feedback:")
    print(f"- {data['phrase']}")
    print(f"- {data['special']}")
else:
    print("❌ Invalid section name! Please choose from M1, F1, O1, or O2.")# ----------------------------------------------------

# ----------------------------------------------------


```
